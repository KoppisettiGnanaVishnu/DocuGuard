```python
#dictionary stores data as key value pairs,a key is used to identify the value

student={
    "name":"Vishnu",
    "age":20,
    "course":'CSE'
}

print(student["name"])
print(student["age"])
print(student["course"])

student["address"]="Delhi"
print(student)

student["age"]=21
print(student)

student.pop("address")
print(student)

if "age" in student:
    print("age exists")
    
for key in student:
    print(key)
    
print(student.keys())
print(student.values())
```
